```python
cap = cv2.VideoCapture(video_name)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

print(f"FPS: {fps}")
print(f"Resolution: {width} x {height}")
```

```python
frame_no = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    image = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

    text = "person. car. bus. truck. bicycle. motorcycle."

    inputs = processor(
        images=image,
        text=text,
        return_tensors="pt"
    )

    with torch.no_grad():
        outputs = model(**inputs)

    results = processor.post_process_grounded_object_detection(
        outputs=outputs,
        input_ids=inputs.input_ids,
        threshold=0.35,
        text_threshold=0.25,
        target_sizes=[image.size[::-1]]
    )

    boxes = results[0]["boxes"]
    labels = results[0]["labels"]
    scores = results[0]["scores"]

    frame = np.array(image)

    for box, label, score in zip(boxes, labels, scores):

        x1, y1, x2, y2 = box.tolist()

        cv2.rectangle(
            frame,
            (int(x1), int(y1)),
            (int(x2), int(y2)),
            (0,255,0),
            2
        )

        cv2.putText(
            frame,
            f"{label} {score:.2f}",
            (int(x1), int(y1)-5),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.5,
            (255,0,0),
            2
        )

    cv2.imwrite(f"output_frames/frame_{frame_no:05d}.jpg",
                cv2.cvtColor(frame, cv2.COLOR_RGB2BGR))

    frame_no += 1

cap.release()

print("Done!")
```

```python
import cv2
import torch
import numpy as np
import supervision as sv

from PIL import Image
from transformers import (
    AutoProcessor,
    AutoModelForZeroShotObjectDetection
)
```

```python
model_id = "IDEA-Research/grounding-dino-base"

processor = AutoProcessor.from_pretrained(model_id)

model = AutoModelForZeroShotObjectDetection.from_pretrained(model_id)
```

```python
tracker = sv.ByteTrack()
```

```python
video_path = "your_video.mp4"

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)

width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
```

```python
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    "output.mp4",
    fourcc,
    fps,
    (width,height)
)
```

```python
TEXT_PROMPT = "person. car. bicycle. bus. truck. motorcycle."
```

```python
while True:

    ret, frame = cap.read()

    if not ret:
        break

    image = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))

    inputs = processor(
        images=image,
        text=TEXT_PROMPT,
        return_tensors="pt"
    )

    with torch.no_grad():
        outputs = model(**inputs)

    results = processor.post_process_grounded_object_detection(
        outputs=outputs,
        input_ids=inputs.input_ids,
        threshold=0.35,
        text_threshold=0.25,
        target_sizes=[image.size[::-1]]
    )

    boxes = results[0]["boxes"].cpu().numpy()
    scores = results[0]["scores"].cpu().numpy()
    labels = results[0]["labels"]

    detections = sv.Detections(
        xyxy=boxes,
        confidence=scores,
        class_id=np.arange(len(boxes))
    )

    detections = tracker.update_with_detections(detections)

    for box, track_id, label in zip(
            detections.xyxy,
            detections.tracker_id,
            labels):

        x1,y1,x2,y2 = map(int,box)

        cv2.rectangle(frame,(x1,y1),(x2,y2),(0,255,0),2)

        cv2.putText(
            frame,
            f"{label} ID:{track_id}",
            (x1,y1-10),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (0,0,255),
            2
        )

    out.write(frame)
```

```python
cap.release()
out.release()

print("Finished!")
```
